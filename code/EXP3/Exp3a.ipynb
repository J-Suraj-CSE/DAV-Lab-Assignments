```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis

# Load the dataset
df = pd.read_csv("diabetes.csv")

# Display first 5 rows
print("Dataset Sample:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
df.info()

# Display summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Numerical columns in the Pima Diabetes dataset
numerical_columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

# Function for univariate analysis
def univariate_analysis(df, columns):
    stats = {}

    for col in columns:
        stats[col] = {
            "Mean": df[col].mean(),
            "Median": df[col].median(),
            "Mode": df[col].mode()[0],
            "Variance": df[col].var(),
            "Standard Deviation": df[col].std(),
            "Skewness": skew(df[col], nan_policy="omit"),
            "Kurtosis": kurtosis(df[col], nan_policy="omit")
        }

    return pd.DataFrame(stats).T

# Perform analysis
result = univariate_analysis(df, numerical_columns)

# Display results
print("\nUnivariate Analysis Results:")
print(result)
```
