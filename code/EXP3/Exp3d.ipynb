```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import r2_score, accuracy_score

# ---------------------------------------
# Load Dataset
# ---------------------------------------

df = pd.read_csv("diabetes.csv")

print("First 5 Rows:")
print(df.head())

print("\nSummary Statistics:")
print(df.describe())

# =======================================
# Linear Regression
# =======================================

X_linear = df[["Glucose", "BloodPressure", "Age"]]
y_linear = df["BMI"]

X_train, X_test, y_train, y_test = train_test_split(
    X_linear,
    y_linear,
    test_size=0.20,
    random_state=42
)

linear_model = LinearRegression()

linear_model.fit(X_train, y_train)

y_pred = linear_model.predict(X_test)

r2 = r2_score(y_test, y_pred)

# =======================================
# Logistic Regression
# =======================================

X_logistic = df[["Glucose", "BloodPressure", "BMI", "Age"]]
y_logistic = df["Outcome"]

X_train, X_test, y_train, y_test = train_test_split(
    X_logistic,
    y_logistic,
    test_size=0.20,
    random_state=42
)

logistic_model = LogisticRegression(max_iter=1000)

logistic_model.fit(X_train, y_train)

prediction = logistic_model.predict(X_test)

accuracy = accuracy_score(y_test, prediction)

# =======================================
# Display Results
# =======================================

print("\n========== MODEL COMPARISON ==========")

print("\nLinear Regression")
print("-----------------------")
print("R² Score :", round(r2, 4))

print("\nLogistic Regression")
print("-----------------------")
print("Accuracy :", round(accuracy * 100, 2), "%")

print("\nConclusion:")
print("Linear Regression predicts BMI using Glucose, BloodPressure and Age.")
print("Logistic Regression predicts whether a patient has diabetes (Outcome).")
```
