```python
import pandas as pd

# Read CSV
text_df = pd.read_csv("data.csv")

# Read Excel
excel_df = pd.read_excel("data.xlsx")

# Read data from web
web_df = pd.read_csv(
    "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
)

# Display
print("CSV Data")
print(text_df.head())

print("\nExcel Data")
print(excel_df.head())

print("\nWeb Data")
print(web_df.head())

# Handle missing values
text_df.ffill(inplace=True)
excel_df.bfill(inplace=True)
web_df.dropna(inplace=True)

# Save processed data
text_df.to_csv("processed_text.csv", index=False)
excel_df.to_excel("processed_excel.xlsx", index=False)

print("\nFiles saved successfully.")
```
