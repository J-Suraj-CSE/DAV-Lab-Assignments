```python
import pandas as pd
from statsmodels.stats.weightstats import ztest

# ---------------------------------------
# Load Dataset
# ---------------------------------------

df = pd.read_csv("diabetes.csv")

# Display first few rows
print("First 5 Rows:")
print(df.head())

# ---------------------------------------
# Perform Z-Test
# H0: Mean Glucose = 100
# H1: Mean Glucose ≠ 100
# ---------------------------------------

z_stat, p_value = ztest(df["Glucose"], value=100)

# ---------------------------------------
# Display Results
# ---------------------------------------

print("\nZ-Test Results")
print("---------------------------")
print("Z-Statistic :", round(z_stat, 4))
print("P-Value     :", round(p_value, 4))

# ---------------------------------------
# Interpretation
# ---------------------------------------

alpha = 0.05

if p_value < alpha:
    print("\nDecision: Reject the Null Hypothesis")
    print("Conclusion: The mean Glucose level is significantly different from 100.")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis")
    print("Conclusion: The mean Glucose level is not significantly different from 100.")
```
