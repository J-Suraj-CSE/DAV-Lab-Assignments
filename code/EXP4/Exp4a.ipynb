```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

# Load Dataset
df = pd.read_csv("diabetes.csv")

# Create Figure
plt.figure(figsize=(12, 5))

# -----------------------------
# Normal Curve for Glucose
# -----------------------------
plt.subplot(1, 2, 1)

sns.histplot(df["Glucose"], kde=True, stat="density", color="skyblue")

x = np.linspace(df["Glucose"].min(), df["Glucose"].max(), 100)

plt.plot(
    x,
    norm.pdf(x, df["Glucose"].mean(), df["Glucose"].std()),
    color="red",
    linewidth=2,
    label="Normal Curve"
)

plt.title("Normal Distribution - Glucose")
plt.xlabel("Glucose")
plt.ylabel("Density")
plt.legend()

# -----------------------------
# Normal Curve for BMI
# -----------------------------
plt.subplot(1, 2, 2)

sns.histplot(df["BMI"], kde=True, stat="density", color="lightgreen")

x = np.linspace(df["BMI"].min(), df["BMI"].max(), 100)

plt.plot(
    x,
    norm.pdf(x, df["BMI"].mean(), df["BMI"].std()),
    color="red",
    linewidth=2,
    label="Normal Curve"
)

plt.title("Normal Distribution - BMI")
plt.xlabel("BMI")
plt.ylabel("Density")
plt.legend()

plt.tight_layout()
plt.show()
```
